# FASE 2.2 :  DATA CLEANING

```python
from thefuzz import process

# 1. Memangkas spasi berlebih (depan/belakang) pada kolom teks
df['Shipping_City'] = df['Shipping_City'].str.strip()

# --- MENGGUNAKAN FUZZY MATCHING SEBAGAI PENGGANTI KAMUS ---
# Daftar Master Kota yang Benar (Bisa ditambah jika dirasa kurang)
kota_master = [
    'Ambon', 'Badung', 'Balikpapan', 'Bandar Lampung', 'Bandung', 'Banjarmasin', 
    'Batam', 'Bekasi', 'Bogor', 'Cirebon', 'Denpasar', 'Depok', 'Jakarta Barat', 
    'Jakarta Pusat', 'Jakarta Selatan', 'Jakarta Timur', 'Jakarta Utara', 
    'Jayapura', 'Jember', 'Kediri', 'Makassar', 'Malang', 'Manado', 'Mataram', 
    'Medan', 'Palembang', 'Pekanbaru', 'Pontianak', 'Purwokerto', 'Semarang', 
    'Solo', 'Sukabumi', 'Surabaya', 'Tangerang', 'Tangerang Selatan', 'Yogyakarta'
]

# Fungsi otomatis untuk mencari kemiripan
def fix_city_fuzzy(city_name):
    best_match, score = process.extractOne(city_name, kota_master)
    if score >= 80: # Jika mirip 80% atau lebih, ubah ke nama master
        return best_match
    else:
        return city_name

# Menerapkan fungsi fuzzy ke kolom kota
df['Shipping_City'] = df['Shipping_City'].apply(fix_city_fuzzy)

# Cek kembali daftar kota untuk memastikan sudah rapi
print("=== DAFTAR KOTA SETELAH FUZZY MATCHING ===")
print(sorted(df['Shipping_City'].unique()))

# ---------------------------------------------------------------

# 2. Menangani nilai kosong (NaN)
# Jika diskon kosong, asumsikan tidak ada diskon (diisi angka 0)
df['Discount_Amount'] = df['Discount_Amount'].fillna(0)
# Jika metode pembayaran kosong, kategorikan sebagai 'Lainnya'
df['Payment_Method'] = df['Payment_Method'].fillna('Lainnya')

# 3. Menangani angka minus (Typo sistem)
# Diasumsikan tanda minus adalah kesalahan ketik, jadi ubah jadi positif (absolut)
df['Unit_Price'] = df['Unit_Price'].abs()
df['Quantity'] = df['Quantity'].abs()
```

# FASE 3: FEATURE ENGINEERING & EXPORT (MEMBUAT METRIK)

```python
# 1. Membuat kolom Total Penjualan Bersih per transaksi
df['Net_Sales'] = (df['Unit_Price'] * df['Quantity']) - df['Discount_Amount']

# 2. Mengekstrak waktu agar mudah divisualisasikan di Power BI
df['Nama_Bulan'] = df['Transaction_Date'].dt.month_name()
df['Nama_Hari'] = df['Transaction_Date'].dt.day_name()
df['Jam_Transaksi'] = df['Transaction_Date'].dt.hour

# 3. Menyimpan hasil akhir menjadi file CSV baru yang bersih
df.to_csv('ecommerce_cleaned_ready.csv', index=False)

print("Pembersihan selesai! Data siap divisualisasikan.")
print("\nContoh 3 baris data terbaru:")
print(df[['Transaction_Date', 'Unit_Price', 'Quantity', 'Net_Sales', 'Nama_Bulan']].head(3))
```
